## <span style="color:#1a237e">📘 Introduction</span>

Exam results are often seen as a reflection of a student's intelligence alone, but in reality
they are shaped by <span style="color:#00838f"><b>many factors working together</b></span>: how consistently a student
attends class, how many hours they study, how well they sleep, how much time they spend on screens,
how stressed or motivated they feel, and even the economic and educational background of their family.

Understanding these factors helps teachers, parents and institutions move from
<span style="color:#e65100"><b>guesswork to evidence</b></span> when supporting students. If we can identify which habits
and conditions are most closely linked to passing or failing, we can design better interventions such as
targeted tutoring, stress-management programmes or study-skills workshops.

```python
# Sleep cannot overlap with studying or online learning,
# so each of these totals must be 24 hours or less
study_sleep = df['study_hours_per_day'] + df['sleep_hours']
online_sleep = df['online_learning_hours'] + df['sleep_hours']

print("Study + sleep over 24 hours:", (study_sleep > 24).sum())
print("Online learning + sleep over 24 hours:", (online_sleep > 24).sum())
```

A day has 24 hours, and a student can't sleep while studying or learning online. So study + sleep, and online learning + sleep, must each be 24 hours or less. Study and online learning weren't added together, because study hours might already include online learning. That would count the same hours twice. The check found 1 student whose online learning + sleep went over 24, so that row is an error.

```python
df = df[(study_sleep <= 24) & (online_sleep <= 24)].copy()
print("Shape after removing impossible rows:", df.shape)
```

Since it's only 1 row out of 100,000, removing it doesn't affect the analysis but makes the data more accurate by bringing the row count to 99,999

```python
Q1 = df['study_hours_per_day'].quantile(0.25)
Q3 = df['study_hours_per_day'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['study_hours_per_day'] < lower) | (df['study_hours_per_day'] > upper)]

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Number of outliers:", len(outliers))
print("Highest study hours:", df['study_hours_per_day'].max())
```

IQR is the range of the middle 50% of students. Anything more than 1.5 × IQR beyond that range is called an outlier. For study hours, the upper limit is about 7.2 hours, and 2,587 students are above it, up to 12 hours. These are unusual but possible, since some students really study that much.Thats why outlier havent been removed.

```python
# Pass = 1, Fail = 0 (needed to calculate pass rates)
df['pass_flag'] = df['pass_status'].map({'Pass': 1, 'Fail': 0}).copy()

# Changing 0/1 into No/Yes so it is easier to read
df['private_tuition'] = df['private_tuition'].map({1: 'Yes', 0: 'No'})
df['internet_access'] = df['internet_access'].map({1: 'Yes', 0: 'No'})
```

Changing 0/1 to No/Yes in tuition and internet access makes tables and charts readable without remembering what 0 and 1 mean.

```python
# Put numbers into groups
df['age_group'] = pd.cut(df['age'], bins=[13, 15, 17, 20],
                         labels=['14-15', '16-17', '18-20'])

df['study_group'] = pd.cut(df['study_hours_per_day'], bins=[0, 1, 2, 4, 24],
                           labels=['Up to 1 hr', '1-2 hrs', '2-4 hrs', 'More than 4 hrs'])

df['sleep_group'] = pd.cut(df['sleep_hours'], bins=[0, 6, 8, 24],
                           labels=['6 hrs or less', '6-8 hrs', 'More than 8 hrs'])

df['screen_group'] = pd.cut(df['daily_screen_time'], bins=[0, 3, 6, 24],
                            labels=['Low', 'Moderate', 'High'])

df['attendance_group'] = pd.cut(df['attendance_percentage'], bins=[0, 75, 90, 100],
                                labels=['75% or less', '75-90%', 'Above 90%'])
```

grouped columns by using bins and intervals of variables to combine certain values to a single bin which would help later in creation of bar charts.

```python
# Combined columns

df['academic_score'] = (df['previous_exam_score'] + df['attendance_percentage']
                        + df['assignment_completion_rate']) / 3
df['academic_score'] = df['academic_score'].round(2).copy()

df.head()
```

Academic score = average of exam score, attendance and assignments. this would get into easy analysis of each student by just looking into this column related to their score card academic wise.

```python
# Removing any extra spaces at the start or end of text values, as a safety check
text_columns = ['gender', 'education_level', 'school_type', 'family_income',
                'parent_education', 'urban_rural', 'class_participation',
                'study_method', 'motivation_level', 'exam_difficulty', 'pass_status']

before = df[text_columns].nunique()

for col in text_columns:
    df[col] = df[col].str.strip()

after = df[text_columns].nunique()

print("Number of unique values before and after stripping:")
print(pd.DataFrame({'before': before, 'after': after}))
```

Used str.strip() on all text columns to remove any extra spaces at the start or end of values, as a safety check. The number of unique values stayed the same before and after, and the unique() check showed no differences in capital letters or spelling. So the text data was already clean and consistent, and no further changes are needed.

```python
category_columns = ['gender', 'education_level', 'school_type', 'family_income','class_participation',
                    'parent_education', 'urban_rural','study_method', 'motivation_level', 'exam_difficulty',
                    'pass_status', 'private_tuition', 'internet_access']


for col in category_columns:
    df[col] = df[col].astype('category').copy()

df.info()
```

### Chart 1: Correlation Heatmap
Used corr() to measure how strongly each number column is linked to passing (pass_flag) and to each other. A heatmap shows all the links in one coloured grid.

```python
# Choosing number columns to compare, plus pass_flag (1 = Pass, 0 = Fail)
columns = ['previous_exam_score', 'attendance_percentage', 'assignment_completion_rate',
           'study_hours_per_day', 'sleep_hours', 'daily_screen_time',
           'stress_level', 'exam_anxiety_level', 'pass_flag']

# corr() finds the link between every pair of columns
corr = df[columns].corr()

# illustration through heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(corr,fmt='.2f',annot=True, cmap='RdBu_r')   # annot=True writes the numbers
plt.title('Correlation Heatmap')
plt.show()
```

**Observation:** Previous exam score has the strongest link with passing (0.36, moderate), followed by study hours (0.22). Assignment completion, attendance and sleep have weak positive links (0.12–0.14), while stress has a weak negative link (-0.11), meaning more stressed students pass slightly less often. Screen time shows no link at all (0.003).

Two clear groups of factors appear. First, attendance, assignment completion and study hours are linked to each other (0.37 and 0.42), showing that hard-working habits go together. Second, stress and exam anxiety are strongly linked (0.64), and both are lower for students who sleep more (-0.20 and -0.13). This suggests that good sleep may help reduce stress, and lower stress is linked to passing.

### Chart 2: Bar Graph
Calculated the pass rate for easy, medium and hard exams using groupby. The red dashed line shows the overall pass rate (77%) for comparison.

```python
# Pass rate (%) for each exam difficulty
rate = df.groupby('exam_difficulty', observed=True)['pass_flag'].mean() * 100

# categorising to Easy, Medium, Hard
rate = rate.reindex(['Easy', 'Medium', 'Hard'])
print(rate.round(2))

# illustration through bar chart
plt.figure(figsize=(6, 4))
ax = rate.plot(kind='bar', color='skyblue', edgecolor='black')
plt.axhline(overall, color='red', linestyle='--', label='Overall pass rate')  # overall pass rate line
ax.bar_label(ax.containers[0], fmt='%.1f%%')   # labels the % on each bar

plt.title('Pass Rate by Exam Difficulty')
plt.ylabel('Pass rate (%)')
plt.ylim(0, 100)
plt.legend()
plt.xticks(rotation=0)
plt.show()
```

**Observation:** When the exam is easy, 90.2% of students pass. For medium exams it drops to 78.9%, and for hard exams only 61.6% pass. So the harder the exam, the more students fail. Out of everything I checked, exam difficulty makes the biggest difference in the pass status

### Chart 3: Heat map
Used a pivot table to find the pass rate for every mix of exam difficulty and study hours, shown as a heatmap. This checks whether studying more helps, especially for hard exams.

```python
# Rows = exam difficulty, columns = study-hours group, boxes = pass rate (%)
pivot = pd.pivot_table(df, values='pass_flag', index='exam_difficulty',
                       columns='study_group', aggfunc='mean', observed=True) * 100

# rows in order wise: Easy, Medium, Hard
pivot = pivot.reindex(['Easy', 'Medium', 'Hard'])
print(pivot.round(2))

#(darker green = higher pass rate)
plt.figure(figsize=(9, 4))
sns.heatmap(pivot, annot=True, fmt='.1f', cmap='YlGn')
plt.title('Pass Rate (%) by Exam Difficulty and Study Hours')
plt.xlabel('Study hours per day')
plt.ylabel('Exam difficulty')
plt.show()
```

**Observation:** In every row, the pass rate goes up as study hours go up, so studying more helps for all exams. The biggest difference is for hard exams. Students who study up to 1 hour pass only 41.5% of the time, but students who study more than 4 hours pass 81.0% of the time, almost double.also noticed that a student who studies more than 4 hours for a hard exam (81.0%) does better than a student who studies up to 1 hour for an easy exam (79.3%). This shows that putting in more study time can make up for a hard exam.

## Methods Used

**pass_flag:** Pass was changed to 1 and Fail to 0. So the average of pass_flag gives the pass rate. Multiplying by 100 shows the pass rate as a percentage.

**groupby():** Used to compare one factor with the pass rate. For example, students were split into attendance groups and the pass rate of each group was found. This allows groups to be compared even when they have different numbers of students.

**pivot_table():** Used to check two factors together, like exam difficulty and study hours. One factor goes in the rows, the other in the columns, and each box shows the pass rate for that combination.

**Heatmap:** A pivot table is just a table of numbers, which is hard to read quickly. A heatmap adds colours to the table, so the highest and lowest pass rates are easy to spot.

**crosstab():** Used to count how many students are in each group, and to get percentages where each row adds up to 100%. This helps compare groups of different sizes fairly and check that no group is too small.

**corr():** Used to see how strongly two number columns are linked. Values go from -1 to +1. Closer to +1  means a stronger link, and closer to 0 means a weaker link.

**Choice of charts:**
- Bar charts: to compare pass rates between groups
- Line charts: to show how the pass rate goes up or down
- Donut charts with subplots: to show the Pass and Fail share, which adds up to 100%
- Boxplots: to compare stress and anxiety level of students and its concentrations.Also to compare passed & failed students with academic_score.
- Histograms: to see how one column is spread out
- Heatmap: to show correlations and pivot tables as coloured grids
- Scatterplot: to show the dispersion of 2 variables.
- swarmplot: to represent each student pass status with a dot.

## <span style="color:#2e7d32">✅ Recommendations</span>

1. **Support low scorers early:** Students who scored below 60 before should be identified at the start and should be given extra classes or mentoring.
2. **Prepare students for hard exams:** Warn students in advance about hard exams and encourage more study time before them.
3. **Track attendance and assignments:** Give quick support to students below 75% attendance or 50% assignment completion.Institutions can also set clear rules, such as a minimum attendance requirement and regular assignment deadlines, to prevent students from falling behind.
